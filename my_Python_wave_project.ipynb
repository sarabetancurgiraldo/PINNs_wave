```python
# @title Neural Network definition

class FCN(nn.Module): #FCN is a class that inherits from nn.Module, which is the base class for all neural network modules in PyTorch.
#The nn.Module class provides a way to define and manage neural network layers.

    "Defines a standard fully-connected network in PyTorch"
    #It defines a very standard architecture of a normal neural network

    #------INITIALIZATION METHOD---------

    def __init__(self, N_INPUT, N_OUTPUT, N_HIDDEN, N_LAYERS):  #__init__ is the constructor method for the class.
      # N_INPUT:  Number of input features.
      # N_OUTPUT: Number of output features.
      # N_HIDDEN: Number of hidden units in each hidden layer.
      # N_LAYERS: Number of hidden layers.

        super().__init__() #calls the constructor of the parent class (nn.Module)

        # activation = nn.ReLU #!!! TG activation function
        # activation = nn.Tanh
        # activation = nn.ELU
        activation = nn.SiLU


        #------ INPUT LAYER ----------
        self.fcs = nn.Sequential(*[ #self.fcs is a sequential container (nn.Sequential) that contains a linear layer (nn.Linear) and an activation function.
                        nn.Linear(N_INPUT, N_HIDDEN), #N_INPUT input features and N_HIDDEN output features
                        activation()])
        #------ HIDDEN LAYERS ---------
        self.fch = nn.Sequential(*[  #another sequential container (nn.Sequential) containing multiple hidden layers.
                        nn.Sequential(*[
                            nn.Linear(N_HIDDEN, N_HIDDEN),
                            activation()]) for _ in range(N_LAYERS-1)])
        #------ OUTPUT LAYER ---------
        self.fce = nn.Linear(N_HIDDEN, N_OUTPUT)

    # ------ FORWARD METHOD ---------
    def forward(self, x): # it specifies how the input data flows through the network layers
        x = self.fcs(x) # x is the input tensor that passes through the input layer ...
        x = self.fch(x) # ... the hidden layers ...
        x = self.fce(x) # ... the output layer
        return x
```
